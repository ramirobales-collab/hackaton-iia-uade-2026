# M03 · Agente antifraude
Cambiar los tres valores y ejecutar la celda.


```python
import math

# Cargar el caso nuevo. El contracargo no se pide: no existe antes de despachar.
monto = 300        # USD
antiguedad = 50    # dias
distancia = 900    # km

z = -2.82 + 0.76 * distancia / 100 - 0.67 * antiguedad / 100
p = 1 / (1 + math.exp(-z))
costo_fn = 0.70 * monto + 20
costo_fp = 0.30 * monto + 10
despachar = p * costo_fn
frenar = (1 - p) * costo_fp

print(f"P(fraude) = {p:.3f}")
print(f"Costo esperado de DESPACHAR = {p:.3f} x {costo_fn:.1f} = {despachar:.1f} USD")
print(f"Costo esperado de FRENAR    = {1 - p:.3f} x {costo_fp:.1f} = {frenar:.1f} USD")
print("Respuesta:", "SI -> FRENAR" if frenar < despachar else "NO -> DESPACHAR")
```
